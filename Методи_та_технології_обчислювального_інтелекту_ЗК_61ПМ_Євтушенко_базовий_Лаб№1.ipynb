# Завдання 1a. Сортування словника за значеннями

> Написати програму на Python, яка сортує словник за значеннями.

````python
data = {'глюкоза': 5.4, 'пульс': 72, 'температура': 36.6, 'сатурація': 98}

asc = dict(sorted(data.items(), key=lambda kv: kv[1]))
desc = dict(sorted(data.items(), key=lambda kv: kv[1], reverse=True))

print("Початковий:     ", data)
print("За зростанням:  ", asc)
print("За спаданням:   ", desc)
````

# Завдання 1c. Об'єднання словників

> Поєднує наступні словники та створює новий
>
> dic1={1:10, 2:20}
>
> dic2={3:30, 4:40}
>
> dic3={5:50, 6:60}
>
> Результат: {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}

````python
dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

merged = {**dic1, **dic2, **dic3}
print("Результат:", merged)

merged2 = {}
for dic in (dic1, dic2, dic3):
    merged2.update(dic)
print("Перевірка:", merged2 == merged)
````

# Завдання 1l. Видалення елемента, якщо він є в множині

> Видаляє елемент якщо він є в set.

````python
s = {1, 2, 3, 4}

for x in (2, 99):
    if x in s:
        s.remove(x)
        print(f"{x} видалено")
    else:
        print(f"{x} у множині немає, пропускаємо")

s.discard(99)
print(s)
````

Метод `discard()` робить те саме, що перевірка з `remove()`, але без помилки.

# Завдання 1q. Сортування кортежів за останнім елементом

> яка отримує список, відсортований у порядку зростання за останнім елементом у кожному кортежі із заданого списку не порожніх кортежів.
>
> Приклад: [(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]
>
> Результат: [(2, 1), (1, 2), (2, 3), (4, 4), (2, 5)]

````python
def sort_by_last(tuples):
    return sorted(tuples, key=lambda t: t[-1])

data = [(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]
print(sort_by_last(data))
````

Ключ `t[-1]` бере останній елемент кортежу, тому сортування йде саме за ним.

# Завдання 1u. Факторіал числа

> Розраховує факторіал числа.

````python
def factorial(n):
    if n < 0:
        raise ValueError("факторіал визначений лише для n >= 0")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

for n in (0, 1, 5, 10):
    print(f"{n}! = {factorial(n)}")

print("Перевірка з math.factorial(5):", math.factorial(5) == factorial(5))
````

# Завдання 1v. Підрахунок великих і малих літер

> Приймає строку та підраховує кількість великих та маленьких букв.
>
> Ввід: 'The quick Brow Fox'
>
> Результат: No. of Upper case characters : 3 No. of Lower case Characters : 12

````python
def count_case(s):
    upper = sum(1 for ch in s if ch.isupper())
    lower = sum(1 for ch in s if ch.islower())
    return upper, lower

text = 'The quick Brow Fox'
up, low = count_case(text)
print("No. of Upper case characters :", up)
print("No. of Lower case Characters :", low)
````

Пробіли не рахуються ні як великі, ні як малі літери, тому в сумі виходить 15, а не 18.

# Завдання 1w. Друк патерну із зірочок

> Друкує наступний патерн, використовуючи цикли
>
> ```
> *
>  *
>  * *
>  * * *
>  * * * *
>  * * *
>  * *
>  *
> *
> ```

````python
n = 4

print('*')
for i in range(1, n + 1):
    print(' ' + '* ' * i)
for i in range(n - 1, 0, -1):
    print(' ' + '* ' * i)
print('*')
````

Патерн відтворено так, як він виглядає в завданні: перший і останній рядки складаються з однієї зірочки без відступу, а проміжні мають відступ в один пробіл.

# Контрольні питання

**1. Що таке словник?**

Словник (dict) зберігає пари «ключ — значення». Значення дістають за ключем, наприклад `d['cat']`, а ключем може бути лише незмінний тип: число, рядок чи кортеж. Словник можна змінювати: додавати, видаляти й оновлювати елементи. Для безпечного доступу є `d.get(key, default)`, а перебирати його зручно через `d.items()`.

**2. Що таке sets?**

Set, або набір, це невпорядкована колекція унікальних елементів. Повторне додавання того самого елемента нічого не змінює, а порядок елементів не гарантується, тому припускати його не можна. Елементи додають через `add`, видаляють через `remove`, перевіряють наявність через `in`. Над множинами можна виконувати об'єднання та перетин, а створювати їх можна й через set comprehension.

**3. Що таке кортеж?**

Кортеж (tuple) це незмінний упорядкований список значень, наприклад `t = (5, 6)`. Він схожий на список, але після створення його не можна змінити. Через цю незмінність кортежі можна використовувати як ключі словників і як елементи наборів, чого списки не дозволяють.
