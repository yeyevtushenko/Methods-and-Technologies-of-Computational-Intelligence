# Імпорт бібліотек

```python
import torch
```

# Завдання 1. Тензор зі списку list(range(9))

> Створіть із списку list(range(9)) тензор. Передбачте та перевірте, який розмір, зсув та кроки він має.

```python
a = torch.tensor(list(range(9)))
print(a)

expected_size = torch.Size([9])
expected_offset = 0
expected_stride = (1,)

print("Розмір:", a.size(), "| передбачено:", expected_size, "| збігається:", a.size() == expected_size)
print("Зсув:  ", a.storage_offset(), "| передбачено:", expected_offset, "| збігається:", a.storage_offset() == expected_offset)
print("Крок:  ", a.stride(), "| передбачено:", expected_stride, "| збігається:", a.stride() == expected_stride)
```

Тензор одновимірний із 9 елементів, тому його розмір `[9]`. Він займає сховище з самого початку, тож зсув дорівнює 0, а щоб перейти до наступного елемента, треба пропустити один елемент сховища, тому крок `(1,)`.

# Завдання 2. Тензор b = a.view(3, 3)

> Створіть тензор b = a.view(3, 3). Яке значення має b[1,1]?

```python
b = a.view(3, 3)
print(b)
print("b[1,1] =", b[1, 1])
print("Спільне сховище:", b.untyped_storage().data_ptr() == a.untyped_storage().data_ptr())
```

Метод `view` лише змінює форму, не копіюючи дані, тому `b` дивиться на те саме сховище, що й `a`. Елемент `b[1,1]` розташований у сховищі за індексом `0 + 3*1 + 1*1 = 4`, тож його значення дорівнює 4.

# Завдання 3. Тензор c = b[1:,1:]

> Створіть тензор c = b[1:,1:]. Передбачте та перевірте, який розмір, зсув та кроки він має.

```python
c = b[1:, 1:]
print(c)

expected_size = torch.Size([2, 2])
expected_offset = 1 * 3 + 1 * 1
expected_stride = (3, 1)

print("Розмір:", c.size(), "| передбачено:", expected_size, "| збігається:", c.size() == expected_size)
print("Зсув:  ", c.storage_offset(), "| передбачено:", expected_offset, "| збігається:", c.storage_offset() == expected_offset)
print("Крок:  ", c.stride(), "| передбачено:", expected_stride, "| збігається:", c.stride() == expected_stride)
```

Зріз бере рядки з 1-го і стовпці з 1-го, тому розмір `[2, 2]`. Перший елемент зрізу `b[1,1]` лежить у сховищі за індексом 4, тож зсув 4. Крок лишається таким самим, як у `b`, тобто `(3, 1)`, бо тензор і далі дивиться на те саме сховище.

# Завдання 4. Математична операція у torch

> Виберіть математичну операцію, таку як косинус або квадратний корінь. Чи можете ви знайти відповідну функцію в бібліотеці torch?

```python
x = torch.tensor([0.0, 1.0, 4.0, 9.0])

print("torch.cos:", torch.cos(x))
print("torch.sqrt:", torch.sqrt(x))
print("Функції є в torch:", hasattr(torch, 'cos'), hasattr(torch, 'sqrt'))
print("x не змінився:", x)
```

Так, відповідні функції є: `torch.cos` для косинуса й `torch.sqrt` для квадратного кореня. Вони повертають новий тензор і не змінюють вхідний.

# Завдання 5. In-place версія функції

> Чи існує версія вашої функції, яка працює in-place?

```python
y = torch.tensor([0.0, 1.0, 4.0, 9.0])

print("Методи з підкресленням є:", hasattr(torch.Tensor, 'cos_'), hasattr(torch.Tensor, 'sqrt_'))

result = y.sqrt_()
print("Після y.sqrt_():", y)
print("Той самий об'єкт:", result is y)

z = torch.tensor([0.0, 1.0, 2.0])
z.cos_()
print("Після z.cos_():", z)
```

Так, in-place версії існують і позначаються підкресленням в кінці імені: `cos_` та `sqrt_`. Вони змінюють сам тензор і повертають його ж, а не створюють новий.

# Висновки

У цьому практикумі ми навчились створювати тензори PyTorch і розуміти, як вони зберігають дані в пам'яті. Метою завдання було засвоїти принципи роботи бібліотеки PyTorch, і її досягнуто, бо всі п'ять завдань виконано, а передбачені значення розміру, зсуву та кроків збіглися з обчисленими.

**Базовий принцип зберігання даних у PyTorch.** Дані тензора лежать в одновимірному суміжному блоці пам'яті, який називається сховищем (Storage). Сам тензор це лише перегляд цього сховища, який задається трьома величинами: розміром (скільки елементів у кожному вимірі), зсувом (індекс у сховищі, що відповідає першому елементу тензора) та кроком (скільки елементів сховища треба пропустити, щоб перейти до наступного елемента вздовж кожного виміру). Елемент `(i, j)` лежить у сховищі за індексом `зсув + крок[0]*i + крок[1]*j`. Через це операції `view`, зрізи та транспонування не копіюють дані, а створюють новий тензор над тим самим сховищем, і зміна одного тензора змінює інші. Щоб отримати незалежну копію, використовується `clone()`, а для транспонованого тензора суміжну копію дає `contiguous()`.

# Контрольні питання

**1. Що являє із себе бібліотека PyTorch?**

PyTorch це бібліотека на основі Python, яка сприяє побудові моделей глибокого навчання та їх використанню в різних додатках, і водночас науковий пакет обчислень. Вона орієнтована на два пункти: заміну NumPy з можливістю використовувати потужність графічних процесорів (GPU) та глибоку дослідницьку платформу, що забезпечує максимальну гнучкість і швидкість. Основною структурою даних у ній є тензор, схожий на масив NumPy, а до важливих модулів належать `torch.nn` для побудови нейронних мереж, `torch.optim` для оновлення ваг, `torch.utils.data` для завантаження даних і `torch.autograd` для автоматичного диференціювання.

**2. Які основні операції з масивами реалізовані у бібліотеці PyTorch?**

Тензори можна створювати через `torch.ones`, `torch.zeros` і `torch.tensor`, дізнаватись їхню форму через `shape` та вказувати тип елементів через `dtype`. Доступ до елементів і зрізи працюють так само, як у NumPy, а перетворення між тензором і масивом NumPy виконують `numpy()` та `torch.from_numpy`. Арифметичні операції (додавання, віднімання, множення, ділення) виконуються поелементно як операторами, так і функціями `torch.add`, `torch.sub`, а матричне множення виконує `torch.mm`. Також є переміщення тензора між CPU і GPU через `to(device)`, зміна форми через `view`, транспонування через `t()` і `transpose`, `contiguous` та `clone`, а методи з підкресленням, наприклад `zero_`, працюють in-place.
